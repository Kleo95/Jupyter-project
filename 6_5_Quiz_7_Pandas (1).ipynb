Use the code below to generate data, and answer the following three questions.
1. Indexing
a) Set "Date" as the index of the DataFrame.
b) Using .loc[], extract the row corresponding to "2024-01-05".
c) Using .iloc[], extract:
rows 0 to 4
columns "Stock" and "Close_Price"
2) Filtering
a) Filter the dataset to include only rows where:

"Stock" is "AAPL"
"Volume" > 5000
b) From the filtered result, keep only:
"Stock", "Close_Price", "Daily_Return_%"

3) Applying Functions
a) Create a new column "Profit_Flag":

"Profit" if "Daily_Return_%" > 0
"Loss" otherwise
(using .apply())
b) Create a column "Price_Category":

"Low" if "Close_Price" < 200
"Medium" if 200 ≤ price ≤ 350
"High" if > 350



```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Number of samples
n = 25

# Generate data
data = pd.DataFrame({
    "Date": pd.date_range(start="2024-01-01", periods=n, freq="D"),
    "Stock": np.random.choice(["AAPL", "GOOG", "TSLA", "AMZN"], size=n),
    "Open_Price": np.round(np.random.uniform(100, 500, size=n), 2),
    "Close_Price": np.round(np.random.uniform(100, 500, size=n), 2),
    "Volume": np.random.randint(1000, 10000, size=n),
    "Sector": np.random.choice(["Tech", "Retail", "Automotive"], size=n)
})

# 1. Indexing
## a) Set "Date" as the index of the DataFrame.
data = data.set_index("Date")

## b) Using .loc[], extract the row corresponding to "2024-01-05".
row_loc = data.loc["2024-01-05"]
print("Row for 2024-01-05:\n", row_loc)

## c) Using .iloc[], extract:
### ---rows 0 to 4
subset_iloc1 = data.iloc[0:5, [0, 2]]
print("\nRows 0 to 4, columns Stock and Close_Price:\n", subset_iloc1)
### ---columns "Stock" and "Close_Price"
subset_iloc1 = data.iloc[0:5, [0, 2]]
print("\nRows 0 to 4, columns Stock and Close_Price:\n", subset_iloc1)

# 2) Filtering
## a) Filter the dataset to include only rows where:
### ---"Stock" is "AAPL"
filtered = data[(data["Stock"] == "AAPL") & (data["Volume"] > 5000)]
### ---"Volume" > 5000
filtered = data[(data["Stock"] == "AAPL") & (data["Volume"] > 5000)]
## b) From the filtered result, keep only:
### ---"Stock", "Close_Price", "Daily_Return_%"
filtered_result = filtered[["Stock", "Close_Price", "Daily_Return_%"]]
print("\nFiltered Result:\n", filtered_result)

# 3) Applying Functions
## a) Create a new column "Profit_Flag":
data["Profit_Flag"] = data["Daily_Return_%"].apply(
    lambda x: "Profit" if x > 0 else "Loss"
)
###---"Profit" if "Daily_Return_%" > 0
###--- "Loss" otherwise
###--- (using .apply())
## b) Create a column "Price_Category":
###---"Low" if "Close_Price" < 200
###---"Medium" if 200 ≤ price ≤ 350
###---"High" if > 350


# b) Price_Category column
def price_category(price):
    if price < 200:
        return "Low"
    elif 200 <= price <= 350:
        return "Medium"
    else:
        return "High"

data["Price_Category"] = data["Close_Price"].apply(price_category)

print("\nFinal DataFrame:\n", data.head())

```
