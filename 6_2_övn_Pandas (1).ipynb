Pandas-övningar

```python
import pandas as pd

df = pd.DataFrame({
    "Namn": ["Anna", "Bo", "Clara", "David", "Eva", "Maja"],
    "Ålder": [25, 31, 29, 45, 22, 53],
    "Lön": [28000, 35000, 32000, 50000, 26000, 50000],
    "Stad": ["Stockholm", "Göteborg", "Malmö", "Stockholm", "Uppsala", "Tumba"]
})

# Nivå 1 – Grunderna
## 1. Visa de första raderna
df.head()

## 2. Visa kolumnen Namn
df["Namn"]

## 3. Hur många rader finns?
len(df)

## 4. Visa information om tabellen
df.info()

## 5. Visa statistik
df.describe()

# Nivå 2 – Filtrering
## 6. Personer äldre än 30
df[df["Ålder"] > 30]

## 7. Personer i Stockholm
df[df["Stad"] == "Stockholm"]

## 8. Lön över 30000
df[df["Lön"] > 30000]

## 9. Under 30 år och lön över 27000
df[(df["Ålder"] < 30) & (df["Lön"] > 27000)]

## 10. Bor inte i Stockholm
df[df["Stad"] != "Stockholm"]

# Nivå 3 – Sortering och analys
## 11. Sortera efter ålder
df.sort_values("Ålder")

## 12. Sortera högsta lön först
df.sort_values("Lön", ascending=False)

## 13. Medelålder
df["Ålder"].mean()

## 14. Högsta lön
df["Lön"].max()

## 15. Lägsta lön
df["Lön"].min()

# Nivå 4 – Skapa kolumner
## 16. Ny kolumn skatt (30%)
df["Skatt"] = df["Lön"] * 0.30

## 17. Lön efter skatt
df["Netto"] = df["Lön"] * 0.70

## 18. Är över 30?
df["Över30"] = df["Ålder"] > 30

# Nivå 5 – Grupp och riktig analys
## 19. Medellön per stad
df.groupby("Stad")["Lön"].mean()

## 20. Antal personer per stad
df["Stad"].value_counts()

# Bonusutmaning
## 21. Vem har högst lön?
df.loc[df["Lön"].idxmax()]

## 22. Vad är snittlönen för personer över 30?
df[df["Ålder"] > 30]["Lön"].mean()


```
