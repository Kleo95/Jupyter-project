```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Download Data
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Step 2: Create a DataFrame
print("DataFrame head:\n", df.head())

# Step 3: Python Functions
def calculate_mean(column):
    return df[column].mean()

print("Mean of Age:", calculate_mean('Age'))

# Step 4: NumPy
age_array = df['Age'].dropna().values  # Drop NaN values for reshaping
reshaped_array = np.reshape(age_array, (len(age_array), 1))
print("Reshaped Age array:\n", reshaped_array)

# Step 5: Pandas
# Using loc()
print("Using loc() to get first 5 rows of Age and Survived:\n", df.loc[:4, ['Age', 'Survived']])

# Using iloc()
print("Using iloc() to get first 5 rows of Age and Survived:\n", df.iloc[:5, [5, 1]])

# Filtering DataFrame
survived = df[df['Survived'] == 1]
print("Filtered DataFrame (Survived):\n", survived.head())

# Step 6: Matplotlib
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

axs[0, 0].hist(df['Age'].dropna(), bins=10, color='blue', alpha=0.7)
axs[0, 0].set_title('Age')

axs[0, 1].hist(df['Fare'], bins=10, color='green', alpha=0.7)
axs[0, 1].set_title('Fare')

axs[1, 0].hist(df['SibSp'], bins=10, color='red', alpha=0.7)
axs[1, 0].set_title('Siblings/Spouses Aboard')

axs[1, 1].hist(df['Parch'], bins=10, color='purple', alpha=0.7)
axs[1, 1].set_title('Parents/Children Aboard')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Download Data
url = "https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv"
df = pd.read_csv(url)

# Step 2: Create a DataFrame
print("DataFrame head:\n", df.head())

# Step 3: Python Functions
def calculate_stats(column):
    mean = df[column].mean()
    median = df[column].median()
    std_dev = df[column].std()
    return mean, median, std_dev

mean_medv, median_medv, std_dev_medv = calculate_stats('medv')
print(f"Mean of medv: {mean_medv}, Median of medv: {median_medv}, Standard Deviation of medv: {std_dev_medv}")

# Step 4: NumPy
medv_array = df['medv'].values
reshaped_array = np.reshape(medv_array, (len(medv_array), 1))
print("Reshaped medv array:\n", reshaped_array)

# Perform a mathematical operation (e.g., normalize the data)
normalized_medv = (reshaped_array - reshaped_array.mean()) / reshaped_array.std()
print("Normalized medv array:\n", normalized_medv)

# Step 5: Pandas
# Using loc()
print("Using loc() to get first 5 rows of medv and chas:\n", df.loc[:4, ['medv', 'chas']])

# Using iloc()
print("Using iloc() to get first 5 rows of medv and chas:\n", df.iloc[:5, [13, 3]])

# Filtering DataFrame
high_medv = df[df['medv'] > 25]
print("Filtered DataFrame (medv > 25):\n", high_medv.head())

# Grouping and calculating mean
chas_grouped = df.groupby('chas').mean()
print("Mean values by chas:\n", chas_grouped)

# Creating a new column for medv categorization
def categorize_medv(medv):
    if medv > 25:
        return 'High'
    elif medv > 15:
        return 'Medium'
    else:
        return 'Low'

df['medv_category'] = df['medv'].apply(categorize_medv)
print("DataFrame with medv_category:\n", df.head())

# Step 6: Matplotlib
fig, axs = plt.subplots(3, 4, figsize=(15, 10))

for i, column in enumerate(df.select_dtypes(include=[np.number]).columns):
    axs[i//4, i%4].hist(df[column], bins=10, color='blue', alpha=0.7)
    axs[i//4, i%4].set_title(column)

plt.tight_layout()
plt.show()

# Scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(df['rm'], df['medv'], alpha=0.5)
plt.title('RM vs MEDV')
plt.xlabel('RM')
plt.ylabel('MEDV')
plt.show()

# Bar plot
plt.figure(figsize=(8, 6))
chas_avg_medv = df.groupby('chas')['medv'].mean()
chas_avg_medv.plot(kind='bar', color='skyblue')
plt.title('Average MEDV by CHAS')
plt.xlabel('CHAS')
plt.ylabel('Average MEDV')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Download Data
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
df = pd.read_csv(url, delimiter=';')

# Step 2: Create a DataFrame
print("DataFrame head:\n", df.head())

# Step 3: Python Functions

# Function to calculate mean, median, and standard deviation
def calculate_stats(df, column):
    mean = df[column].mean()
    median = df[column].median()
    std_dev = df[column].std()
    return mean, median, std_dev

mean_quality, median_quality, std_dev_quality = calculate_stats(df, 'quality')
print(f"Mean of quality: {mean_quality}, Median of quality: {median_quality}, Standard Deviation of quality: {std_dev_quality}")

# Function to reshape data
def reshape_data(df, column):
    array = df[column].values
    reshaped_array = np.reshape(array, (len(array), 1))
    return reshaped_array

reshaped_quality = reshape_data(df, 'quality')
print("Reshaped quality array:\n", reshaped_quality)

# Function to normalize data
def normalize_data(array):
    normalized_array = (array - array.mean()) / array.std()
    return normalized_array

normalized_quality = normalize_data(reshaped_quality)
print("Normalized quality array:\n", normalized_quality)

# Function to filter DataFrame
def filter_dataframe(df, column, condition):
    filtered_df = df[df[column] > condition]
    return filtered_df

high_quality = filter_dataframe(df, 'quality', 6)
print("Filtered DataFrame (quality > 6):\n", high_quality.head())

# Function to group data and calculate mean
def group_and_calculate_mean(df, group_column):
    grouped_df = df.groupby(group_column).mean()
    return grouped_df

alcohol_grouped = group_and_calculate_mean(df, 'alcohol')
print("Mean values by alcohol:\n", alcohol_grouped)

# Function to categorize column
def categorize_column(df, column, high_threshold, medium_threshold):
    def categorize(value):
        if value > high_threshold:
            return 'High'
        elif value > medium_threshold:
            return 'Medium'
        else:
            return 'Low'
    df[f'{column}_category'] = df[column].apply(categorize)
    return df

df = categorize_column(df, 'quality', 7, 5)
print("DataFrame with quality_category:\n", df.head())

# Function to plot histograms
def plot_histograms(df):
    fig, axs = plt.subplots(3, 4, figsize=(15, 10))
    for i, column in enumerate(df.select_dtypes(include=[np.number]).columns):
        axs[i//4, i%4].hist(df[column], bins=10, color='blue', alpha=0.7)
        axs[i//4, i%4].set_title(column)
    plt.tight_layout()
    plt.show()

plot_histograms(df)

# Function to create scatter plot
def plot_scatter(df, x_column, y_column):
    plt.figure(figsize=(8, 6))
    plt.scatter(df[x_column], df[y_column], alpha=0.5)
    plt.title(f'{x_column} vs {y_column}')
    plt.xlabel(x_column)
    plt.ylabel(y_column)
    plt.show()

plot_scatter(df, 'alcohol', 'quality')

# Function to create bar plot
def plot_bar(df, group_column, value_column):
    plt.figure(figsize=(8, 6))
    grouped_avg = df.groupby(group_column)[value_column].mean()
    grouped_avg.plot(kind='bar', color='skyblue')
    plt.title(f'Average {value_column} by {group_column}')
    plt.xlabel(group_column)
    plt.ylabel(f'Average {value_column}')
    plt.show()

plot_bar(df, 'quality_category', 'alcohol')
```
