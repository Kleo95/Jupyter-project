Koden simulerar aktiedata för 100 dagar så man kan öva analys i Pandas.
Explore the attributes and methods we covered in the Pandas tutorial .ipynb file (shared in Module 6) using the data object below

```python
# generating data
import pandas as pd
import numpy as np

# Sätter ett startvärde för slumpen
# Det gör att samma slumpade data skapas varje gång man kör koden.
np.random.seed(42)

# Number of samples (vi vill ha 100 rader data).
n = 100

# Generate data/skapas en tabell
data = pd.DataFrame({
    # Skapar datum från 1 januari 2024 och 100 dagar framåt, varje dag
    "Date": pd.date_range(start="2024-01-01", periods=n, freq="D"),
    # Väljer slumpmässigt aktienamn:AAPL = Apple, GOOG = Google,TSLA = Tesla, AMZN = Amazon
    # 100 slumpade val.
    "Stock": np.random.choice(["AAPL", "GOOG", "TSLA", "AMZN"], size=n),
    # Skapar n slumpade öppnings- och stängningspriser mellan 100 och 500, med 2 decimaler
    "Open_Price": np.round(np.random.uniform(100, 500, size=n), 2),
    "Close_Price": np.round(np.random.uniform(100, 500, size=n), 2),
    # Skapar slumpad handelsvolym (hur många aktier som handlades) mellan 1000 och 9999
    "Volume": np.random.randint(1000, 10000, size=n),
    # Väljer slumpmässigt sektor: Tech = Teknik, Retail = Handel, Automotive = Bilar
    "Sector": np.random.choice(["Tech", "Retail", "Automotive"], size=n)
})

print("Indata\n", data, "\n")
```
